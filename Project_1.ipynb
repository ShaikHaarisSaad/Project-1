```python
x = input('Enter if you have email: ')

while x.lower() != 'yes' and x.lower() != 'no':
    print('Invalid input')
    x = input('Enter if you have email: ')


if x.lower() == 'yes':

    email = input('Enter the email: ')

    while '@' not in email:
        print('Email is not accepted')
        email = input('Enter the email: ')

    print('Email is accepted')


elif x.lower() == 'no':

    name = input('Enter the name: ')
    email = input('Create the email: ')

    while '@' not in email:
        print('Invalid email')
        email = input('Create the email: ')

    n = input('Enter the password: ')

    while True:

        upper = False
        lower = False
        number = False

        for i in n:

            if i.isupper():
                upper = True

            if i.islower():
                lower = True

            if i.isdigit():
                number = True

        if len(n) >= 8 and upper and lower and number:
            print('Accepted password')
            break

        print('Weak password')
        print('Use at least 8 characters, 1 uppercase, 1 lowercase, and 1 number')

        n = input('Enter the password again: ')

    print('Registration successful!')
```
